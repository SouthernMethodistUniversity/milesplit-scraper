```python
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
import re
import math
import time
import random
from io import StringIO
from datetime import date
```

```python
base_url = "https://www.milesplit.com/results/?"
years = [str(x) for x in list(range(2006,2025))]
months = [str(x) for x in list(range(1,12))]
levels = ["hs", "college", "open", "pro"]
```

```python
url = base_url + "month=" + months[0] + "&year=" + years[0] + "&level=" + levels[0]
print(url)
```

```python
r = requests.get(url)
```

```python
soup = BeautifulSoup(r.content, 'html.parser')
```

```python
links = soup.select("td.name a")
```

```python
links[0]['href']
```

```python
r2 = requests.get(links[0]['href'])
```

```python
soup2 = BeautifulSoup(r2.content, 'html.parser')
```

```python
l3 = soup2.select("option")[0]["value"]
```
